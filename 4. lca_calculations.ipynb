```python
def interpolate_and_cumulative(df):
    """
    Interpolates missing years for each (Scenario, Metal, Technology) group,
    and returns:
    - interpolated DataFrame
    - cumulative impact values aggregated across all years

    Parameters
    ----------
    df : pd.DataFrame
        Must contain columns: ['Scenario', 'Year', 'Metal', 'Technology', <impact categories>]

    Returns
    -------
    tuple: (interpolated_df, cumulative_df)
        interpolated_df: pd.DataFrame with interpolated rows
        cumulative_df: pd.DataFrame aggregated over all years
    """
    import pandas as pd

    group_cols = ['Scenario', 'Metal', 'Technology']
    impact_cols = df.columns.difference(['Scenario', 'Year', 'Metal', 'Technology'])
    all_years = list(range(df['Year'].min(), df['Year'].max() + 1))

    def interpolate_group(group):
        group = group.drop_duplicates(subset='Year').set_index('Year').sort_index()
        group = group.reindex(all_years)
        for col in group_cols:
            group[col] = group[col].ffill()
        group[impact_cols] = group[impact_cols].interpolate(method='linear', limit_direction='both')
        return group.reset_index()

    # Interpolate missing years
    df_interp = df.groupby(group_cols, group_keys=False).apply(interpolate_group).reset_index(drop=True)

    # Aggregate across all years to get cumulative impact
    df_cumulative = df_interp.groupby(group_cols)[impact_cols].sum().reset_index()

    return df_interp, df_cumulative
```

```python
# Define the EXACT function that reproduces the output of impact_share_by_group_final_corrected_v2.xlsx

def generate_exact_impact_share_percentages(df_yearly, df_cumulative):
    """
    Compute and return a dictionary of impact share percentages that exactly match the
    structure of the 'impact_share_by_group_final_corrected_v2.xlsx' file.

    Includes:
    - impact_share_by_metal_cumulative
    - impact_share_by_technology_cumulative
    - impact_share_by_metal_technology_cumulative
    - impact_share_by_metal_yearly
    - impact_share_by_technology_yearly
    - metal_technology_yearly
    """
    results = {}
    impact_cols = df_cumulative.columns.difference(['Scenario', 'Year', 'Metal', 'Technology'])

    def share_by_cumulative(df, group_cols, key_name):
        grouped = df.groupby(group_cols)[impact_cols].sum().reset_index()
        total = grouped[impact_cols].sum()
        grouped[impact_cols] = grouped[impact_cols].div(total) * 100
        results[key_name] = grouped

    def share_by_yearly(df, group_cols, key_name):
        grouped = df.groupby(group_cols)[impact_cols].sum().reset_index()
        percent_list = []
        for year, group in grouped.groupby('Year'):
            total = group[impact_cols].sum()
            temp = group.copy()
            temp[impact_cols] = group[impact_cols].div(total) * 100
            percent_list.append(temp)
        results[key_name] = pd.concat(percent_list, ignore_index=True)

    # Cumulative
    share_by_cumulative(df_cumulative, ['Metal'], 'pct_metal_cum')
    share_by_cumulative(df_cumulative, ['Technology'], 'pct_tech_cum')
    share_by_cumulative(df_cumulative, ['Metal', 'Technology'], 'pct_metal_tech_cum')

    # Yearly
    share_by_yearly(df_yearly, ['Year', 'Metal'], 'pct_metal')
    share_by_yearly(df_yearly, ['Year', 'Technology'], 'pct_tech')
    share_by_yearly(df_yearly, ['Year', 'Metal', 'Technology'], 'pct_metal_tech')

    return results

# Run the function on the loaded yearly and cumulative DataFrames
final_exact_results = generate_exact_impact_share_percentages(df_nze_damage_int, df_nze_damage_cum)

# Save to Excel
final_exact_output_path = "results/corrected/df_nze_damage_regionalized_detailed_pct.xlsx"
with pd.ExcelWriter(final_exact_output_path, engine='xlsxwriter') as writer:
    for sheet_name, sheet_df in final_exact_results.items():
        sheet_df.to_excel(writer, sheet_name=sheet_name[:31], index=False)

final_exact_output_path
```

```python
# Define a simplified version that returns only total midpoint contribution, per year and cumulative

def generate_midpoint_total_only(df_yearly, df_cumulative, EQ, HH):
    """
    Compute total midpoint contribution to Total EQ and Total HH,
    shown cumulatively and for each year (aggregated globally).

    Parameters
    ----------
    df_yearly : pd.DataFrame
    df_cumulative : pd.DataFrame
    EQ : list of str
    HH : list of str

    Returns
    -------
    dict of pd.DataFrame
        - midpoint_percent_total_cumulative
        - midpoint_percent_total_yearly
    """
    results = {}

    # Cumulative
    total = df_cumulative.sum(numeric_only=True)
    total_eq = total['Total ecosystem quality']
    total_hh = total['Total human health']
    cumulative_data = []

    for col in EQ:
        cumulative_data.append({'Impact category': col, 'Relative to Total EQ (%)': total[col] / total_eq * 100})
    for col in HH:
        cumulative_data.append({'Impact category': col, 'Relative to Total HH (%)': total[col] / total_hh * 100})

    results['midpoint_percent_total_cumulative'] = pd.DataFrame(cumulative_data)

    # Yearly
    yearly_data = []
    for year, group in df_yearly.groupby("Year"):
        group_total = group.sum(numeric_only=True)
        group_total_eq = group_total['Total ecosystem quality']
        group_total_hh = group_total['Total human health']

        for col in EQ:
            yearly_data.append({
                'Year': year,
                'Impact category': col,
                'Relative to Total EQ (%)': group_total[col] / group_total_eq * 100
            })
        for col in HH:
            yearly_data.append({
                'Year': year,
                'Impact category': col,
                'Relative to Total HH (%)': group_total[col] / group_total_hh * 100
            })

    results['midpoint_percent_total_yearly'] = pd.DataFrame(yearly_data)

    return results

# Run simplified total-only function
midpoint_total_only = generate_midpoint_total_only(df_nze_damage_int, df_nze_damage_cum, EQ, HH)

# Save to Excel
output_total_only_path = "results/corrected/df_nze_damage_regionalized_detailed_midpoint_pct.xlsx"
with pd.ExcelWriter(output_total_only_path, engine='xlsxwriter') as writer:
    for sheet_name, sheet_df in midpoint_total_only.items():
        sheet_df.to_excel(writer, sheet_name=sheet_name[:31], index=False)

output_total_only_path
```
